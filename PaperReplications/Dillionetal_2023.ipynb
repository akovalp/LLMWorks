This work is a replication of the paper "Can AI language models replace human participants?" by Danica Dillion, Niket Tandon, Yuling Gu, and Kurt Gray (2023). 
In the original paper, the authors reviewed evidence on whether large language models (LLMs) like GPT-3 can simulate human judgment and serve as replacements for human participants in psychological research.
The key findings were:
LLMs can produce remarkably human-like moral judgments, with GPT-3.5 achieving a 0.95 correlation with average human ratings across hundreds of moral scenarios.
LLMs have also shown human-like judgments in domains like predicting voting, economic games, problem solving, and classic experiments like the Milgram obedience study.
However, LLMs may be limited in behavioral tasks, topics with complex intuitions, generalizing beyond Western English speakers, and capturing individual differences.
The replication will focus on simulating studies from moral psychology and social psychology using LLMs. 
While API costs may limit large samples, the replication will demonstrate the approach on a smaller scale and allow key findings to be reproduced. Careful prompt design and outcome parsing will be critical to match the original experiments.

```python
def clasify_morality(sentence):
    response = openai.ChatCompletion.create(
        model = "gpt-3.5-turbo-0613",
        messages = [
            {
                "role" : "system",
                "content" : "You are a moral judgment evaluator and you will give an overal score for some moral sentences"
            },
            {
                "role" : "user",
                "content": f""" 
                For each action below, I will rate on a scale of -4 to 4 how unethical or ethical it is. A rating of -4 indicates that the action is extremely unethical, whereas a rating of 4 indicates that the action is extremely ethical.
                ### Examples:
                Person X pushed an amputee in front of a train because the amputee made them feel uncomfortable.
                [Rating]: -3.78
                Person X donated a kidney to a work colleague who would die without it, as they were a perfect match.
                [Rating]: 3.51
                Person X never learned how to ride a bicycle.
                [Rating]: -0.01
                Person X has memorized three of Shakespeare's plays.
                [Rating]: 0.28
                {sentence}
                [Rating]:
                """
            },
        ],
        temperature = 0,
    )
    return response 

for sentence in moral_scenarios:
    sentence = sentence
    response = clasify_morality(sentence)
    score = response['choices'][0]['message']['content']
    print(f"{sentence} \n {score}")
```
